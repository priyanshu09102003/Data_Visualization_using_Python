```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create an array x

x = np.linspace(0, 5 , 11)
y = x**2
```

**Using MatplotLib**

```python
plt.plot(x, y)
plt.title("Basic Plot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
```

**SUBPLOTS IN MATPLOTLIB**
(Needed to create multiple plots in matplotlib)

```python
# Divides the canvas of the matplotlib into rows and cols
plt.subplot(2,2, 1)
plt.subplot(2,2, 2)
plt.subplot(2,2, 3)
plt.subplot(2,2, 4)
plt.plot(x,y)
```

**OBJECT ORIENTED WAY**

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x, y)
axis2 = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axis2.plot(x,y)
```

**TYPES OF PLOTS**

```python
plt.subplot(2,2,1)
plt.plot(x,y) #This is the simple line chart/plot
plt.title("Line Plot")

plt.subplot(2,2,2)
plt.scatter(x,y) #This is the scatter plot --> Used mainly in regression algo visualization
plt.title("Scatter Plot")

plt.subplot(2,2,3)
plt.hist(x) #This is the histogram --> Takes only x axis value
plt.title("Histogram Plot")

plt.subplot(2,2,4)
plt.boxplot(x) #This is the boxplot --> Takes only x axis value
plt.title("Box Plot")
```
