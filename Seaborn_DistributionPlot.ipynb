**DISTRIBUTION PLOTS**

```python
# Importing the library
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("tips") #Loading the demo dataset
df
```

```python
# 1) DISPLOT
plt.subplot(1, 2, 1)
sns.histplot(df["total_bill"])
plt.title("Without KDE")

plt.subplot(1, 2, 2)
sns.histplot(df["total_bill"], kde=True)
plt.title("With KDE")
```

```python
# JOINT PLOT AND TYPES
sns.jointplot(data = df, x = "total_bill", y = "tip", kind = "scatter") #Scatter


sns.jointplot(data = df, x = "total_bill", y = "tip", kind = "kde") #KDE



sns.jointplot(data = df, x = "total_bill", y = "tip", kind = "hex") #HEX



sns.jointplot(data = df, x = "total_bill", y = "tip", kind = "hist") #HIST

sns.jointplot(data = df, x = "total_bill", y = "tip", kind = "reg") #REGRESSION



```

```python
# Pair Plot -> Creates all possible pair combinations

sns.pairplot(df)
sns.pairplot(df, hue='sex')
```
